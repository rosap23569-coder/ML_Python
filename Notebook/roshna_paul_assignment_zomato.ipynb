```python
#Import Libraries
import pandas as pd
import numpy as np
```

```python
#Load Dataset
df = pd.read_csv('/test_zomato.csv')
```

```python
print("Shape of the dataset:", df.shape)
```

```python
print("Column names:")
print(df.columns.tolist())
```

```python
print("Data types:")
print(df.dtypes)
```

```python
print("Missing values");
print(df.isnull().sum())
```

```python
duplicate_count = df.duplicated().sum()
print("Number of duplicate records:",duplicate_count)
```

```python
df[df.duplicated()]
```

```python
print(df['Average Cost for two'].dtype)
```

```python
df['Average Cost for two'] = (
    df['Average Cost for two']
    .astype(str)
    .str.replace(',', '', regex=False)
)
```

```python
df['Average Cost for two'] = pd.to_numeric(
    df['Average Cost for two'],
    errors='coerce'
)
```

```python
print("Basic Statistical Analysis:")
display(df.describe())
```

```python
print("Mean:")
print(df.mean(numeric_only=True))

print("\nMedian:")
print(df.median(numeric_only=True))

print("\nMinimum:")
print(df.min(numeric_only=True))

print("\nMaximum:")
print(df.max(numeric_only=True))

print("\nStandard Deviation:")
print(df.std(numeric_only=True))
```

```python
top_10_cities = df['City'].value_counts().head(10)

print("Top 10 cities with the highest number of restaurants:")
print(top_10_cities)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x=top_10_cities.values,y=top_10_cities.index)
plt.title('Top 10 Cities with Highest Number of Restaurants')
plt.xlabel('Number of Restaurants')
plt.ylabel('City')
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.histplot(df['Aggregate rating'])
plt.title('Distribution of Restaurant Ratings')
plt.xlabel('Aggregate Rating')
plt.ylabel('Number of Restaurants')
plt.show()
```

```python
delivery_counts = df['Has Online delivery'].value_counts()
plt.figure(figsize=(7, 5))
sns.barplot(
    x=delivery_counts.index,
    y=delivery_counts.values
)
plt.title('Restaurants Offering Online Delivery')
plt.xlabel('Online Delivery')
plt.ylabel('Number of Restaurants')
plt.show()
```

```python
cuisine_counts = (
    df['Cuisines']
    .dropna()
    .str.split(',')
    .explode()
    .str.strip()
    .value_counts()
    .head(10)
)
plt.figure(figsize=(10, 6))

sns.barplot(
    x=cuisine_counts.values,
    y=cuisine_counts.index
)

plt.title('Top 10 Most Common Cuisines')
plt.xlabel('Number of Restaurants')
plt.ylabel('Cuisine')
plt.show()
```

```python
avg_rating_price = (
    df.groupby('Price range')['Aggregate rating']
    .mean()
    .reset_index()
)

plt.figure(figsize=(8, 5))

sns.lineplot(
    data=avg_rating_price,
    x='Price range',
    y='Aggregate rating',
    marker='o'
)

plt.title('Average Restaurant Rating by Price Range')
plt.xlabel('Price Range')
plt.ylabel('Average Rating')
plt.xticks(sorted(df['Price range'].dropna().unique()))

plt.show()
```

```python
city_counts = df['City'].value_counts()

print("City with the highest restaurant presence:")
print(city_counts.idxmax())
print("Number of restaurants:", city_counts.max())
#Which city has the highest restaurant presence?: New Delhi
```

```python
cuisine_counts = (
    df['Cuisines']
    .dropna()
    .str.split(',')
    .explode()
    .str.strip()
    .value_counts()
)

print("Most popular cuisine:", cuisine_counts.idxmax())
print("Number of restaurants:", cuisine_counts.max())
#What is the most popular cuisine?: North Indian
```

```python
delivery_ratings = df.groupby(
    'Has Online delivery'
)['Aggregate rating'].mean()

print(delivery_ratings)
#Do restaurants with Online Delivery generally have higher ratings? Restaurants with Online Delivery have higher average ratings.
```

```python
if delivery_ratings['Yes'] > delivery_ratings['No']:
    print("Restaurants with Online Delivery have higher average ratings.")
else:
    print("Restaurants without Online Delivery have higher average ratings.")
```

```python
#Do restaurants offering Table Booking receive better ratings? : Restaurants with Table Booking have higher average ratings.
table_ratings = df.groupby(
    'Has Table booking'
)['Aggregate rating'].mean()

print(table_ratings)
if table_ratings['Yes'] > table_ratings['No']:
    print("Restaurants with Table Booking have higher average ratings.")
else:
    print("Restaurants without Table Booking have higher average ratings.")
```
